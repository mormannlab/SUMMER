# Visualize Movie Frames with Annotations

This notebook demonstrates how to use `visualize_frame_with_annotations` to display a DVD movie frame alongside its NWB annotation labels.

```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path(".").resolve().parent))

%matplotlib inline
```

```python
from movie_wrapper.visualize_movie_frames import visualize_frame_with_annotations
```

```python
# Paths to example data
FRAMES_DIR = Path("input")        # DVD frames: frame_010809.jpg – frame_010839.jpg
DATA_DIR = Path("data")           # NWB files: sub-1.nwb, sub-2.nwb

# Pick an example frame and patient
DVD_FRAME_NUMBER = 10815
PATIENT = "1"
```

```python
fig = visualize_frame_with_annotations(
    frames_dir=FRAMES_DIR,
    dvd_frame_number=DVD_FRAME_NUMBER,
    patient=PATIENT,
    data_dir=DATA_DIR,
)
```

```python
# Try a second patient with the same frame
fig = visualize_frame_with_annotations(
    frames_dir=FRAMES_DIR,
    dvd_frame_number=10820,
    patient="2",
    data_dir=DATA_DIR,
)
```

```python
# Save an example to disk
fig = visualize_frame_with_annotations(
    frames_dir=FRAMES_DIR,
    dvd_frame_number=10830,
    patient="1",
    data_dir=DATA_DIR,
    save_path="output/example_visualization.png",
)
```
